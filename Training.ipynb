```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

model.train(
    data="full_cable/data.yaml",
    epochs=130,
    imgsz=320,
    batch=16,
    patience=20,
    device=0,
    augment=True
)
```

```python
model.val()
```

```python
model.predict(source="full_cable/valid/images", save=True)
```
